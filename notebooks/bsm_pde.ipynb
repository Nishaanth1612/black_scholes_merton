# The Black–Scholes PDE

We now have everything we need:

- The stock equation $dS$.
- Itô's lemma for the derivative $df$.
- The portfolio $\Pi = -f + f_S S$.
- The no-arbitrage condition $d\Pi = r\Pi \, dt$.




**The stock** follows geometric Brownian motion:

$$
dS = \mu S \, dt + \sigma S \, dB_t.
$$

**The derivative** follows Itô's lemma:

$$
df
=
\left(
f_t + \mu S f_S + \tfrac{1}{2} \sigma^2 S^2 f_{SS}
\right) dt
+
\sigma S f_S \, dB_t.
$$

**The portfolio** is short one derivative and long $f_S$ shares:

$$
\Pi = -f + f_S S.
$$

**The change in the portfolio** is therefore

$$
d\Pi = -df + f_S \, dS.
$$

## Substitute $df$ and $dS$

Plug the two equations into $d\Pi$:

$$
d\Pi
=
-\left[
\left(f_t + \mu S f_S + \tfrac{1}{2} \sigma^2 S^2 f_{SS}\right) dt
+ \sigma S f_S \, dB_t
\right]
+
f_S \left[\mu S \, dt + \sigma S \, dB_t\right].
$$

Collect the $dB_t$ terms:

$$
-\sigma S f_S \, dB_t + f_S \cdot \sigma S \, dB_t = 0.
$$

They cancel **exactly**. This is why we chose $f_S$ shares.

Collect the $\mu$ terms:

$$
-\mu S f_S \, dt + f_S \cdot \mu S \, dt = 0.
$$

They also cancel. The drift $\mu$ disappears.

What remains is

$$
{
d\Pi = -\left(f_t + \tfrac{1}{2} \sigma^2 S^2 f_{SS}\right) dt.
}
$$

**No $dB_t$. No $\mu$.** The portfolio is deterministic.

## Apply no-arbitrage

From the riskless-portfolio notebook, a deterministic portfolio must
earn the risk-free rate:

$$
d\Pi = r \Pi \, dt.
$$

Substitute the expressions for $d\Pi$ and $\Pi$:

$$
-\left(f_t + \tfrac{1}{2} \sigma^2 S^2 f_{SS}\right) dt
=
r \left(-f + f_S S\right) dt.
$$

Cancel $dt$ on both sides:

$$
-\left(f_t + \tfrac{1}{2} \sigma^2 S^2 f_{SS}\right)
=
r \left(-f + f_S S\right).
$$

Rearrange:

$$
-\left(f_t + \tfrac{1}{2} \sigma^2 S^2 f_{SS}\right)
=
-rf + r S f_S,
$$

$$
-f_t - \tfrac{1}{2} \sigma^2 S^2 f_{SS}
=
-rf + r S f_S,
$$

$$
f_t + r S f_S + \tfrac{1}{2} \sigma^2 S^2 f_{SS} - r f = 0.
$$

So we obtain the **Black–Scholes PDE**:

$$
{
f_t + r S f_S + \tfrac{1}{2} \sigma^2 S^2 f_{SS} - r f = 0.
}
$$
