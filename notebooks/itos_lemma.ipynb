# Itô's Lemma

*The chain rule for Brownian motion.*

**What confused me:** Newton's chain rule works for every smooth
function. Why can't I just use it for $f(B_t)$? Why does an extra
term appear out of nowhere?

**The key insight that unlocked it:** Newton's chain rule drops the
second-order Taylor term because $(\Delta x)^2 \to 0$ faster than
$\Delta x$. But for Brownian motion, $(\Delta B)^2 \approx \Delta t$,
which is the *same order* as $\Delta t$. So the second-order term
cannot be dropped. It is precisely this surviving term that makes
Itô's lemma different from Newton's rule.

## 1. The ordinary chain rule

In Newton's calculus, if $f$ is a twice-differentiable function and
$x(t)$ is a smooth path, then a small change in $f$ is given by the
Taylor expansion:

$$
\Delta f
=
f'(x) \,\Delta x
+
\frac{1}{2} f''(x) \,(\Delta x)^2
+
\cdots
$$

The chain rule says: for a smooth path, $(\Delta x)^2$ vanishes faster
than $\Delta x$ as $\Delta t \to 0$. So the second-order term is
negligible, and to first order:

$$
df = f'(x) \, dx.
$$

This is the familiar rule. Derivative times differential. No second
term.

### Why does $(\Delta x)^2$ vanish?

For a smooth path, $\Delta x \approx x'(t) \,\Delta t$. So

$$
(\Delta x)^2 \approx (x'(t))^2 (\Delta t)^2.
$$

This is of order $(\Delta t)^2$, while $\Delta x$ is of order
$\Delta t$. Since $(\Delta t)^2 \ll \Delta t$, the second-order term
is smaller and can be dropped.

## 2. Why this fails for Brownian motion

For Brownian motion, the increment over a small time $\Delta t$ has
standard deviation $\sqrt{\Delta t}$. So the typical size of the
increment is

$$
|\Delta B| \sim \sqrt{\Delta t}.
$$

Now square it:

$$
(\Delta B)^2 \sim (\sqrt{\Delta t})^2 = \Delta t.
$$

**The squared increment has the same order as $\Delta t$.** It does
not vanish. It is not negligible.

So when we apply the Taylor expansion to $f(B_t)$, we cannot drop the
second-order term. It contributes at the same order as the first-order
term.

This is the entire difference between Itô's lemma and Newton's chain
rule.

## 3. Deriving Itô's lemma

Let $f(t, x)$ be a twice-differentiable function. We want to compute
$df(t, B_t)$.

Start with the Taylor expansion:

$$
\Delta f
=
\frac{\partial f}{\partial t} \,\Delta t
+
\frac{\partial f}{\partial x} \,\Delta B
+
\frac{1}{2} \frac{\partial^2 f}{\partial x^2} \,(\Delta B)^2
+
\frac{1}{2} \frac{\partial^2 f}{\partial t^2} \,(\Delta t)^2
+
\frac{\partial^2 f}{\partial t \partial x} \,\Delta t \,\Delta B
+
\cdots
$$

Now check the order of each term:

| Term | Order | Keep? |
|---|---|---|
| $\partial_t f \,\Delta t$ | $\Delta t$ | Yes |
| $\partial_x f \,\Delta B$ | $\sqrt{\Delta t}$ | Yes |
| $\frac{1}{2} \partial_{xx} f \,(\Delta B)^2$ | $\Delta t$ | **Yes** |
| $\frac{1}{2} \partial_{tt} f \,(\Delta t)^2$ | $(\Delta t)^2$ | No |
| $\partial_{tx} f \,\Delta t \,\Delta B$ | $(\Delta t)^{3/2}$ | No |

Only three terms survive. Passing to the limit:

$$
df
=
\frac{\partial f}{\partial t} \, dt
+
\frac{\partial f}{\partial x} \, dB
+
\frac{1}{2} \frac{\partial^2 f}{\partial x^2} \,(dB)^2.
$$

Now use the key fact from the Brownian motion notebook:

$$
(dB)^2 = dt.
$$

Substituting:

$$
{
df
=
\left(
\frac{\partial f}{\partial t}
+
\frac{1}{2} \frac{\partial^2 f}{\partial x^2}
\right) dt
+
\frac{\partial f}{\partial x} \, dB
}
$$

This is **Itô's lemma**. The second derivative term is the extra
contribution that Newton's chain rule would miss.

## 4. Specializing Itô's lemma to a stock price

So far, Itô's lemma is stated for a general Brownian motion $B_t$.
But in financial literature, we care about functions of the **stock price** $S_t$.

We have already established that the returns of the stock price follows
geometric Brownian motion:

$$
dS_t = \mu S_t \, dt + \sigma S_t \, dB_t.
$$

Here $\mu$ is the drift and $\sigma$ is the volatility. Compare this
to the general form

$$
dX_t = a_t \, dt + b_t \, dB_t.
$$

We can read off:

$$
a_t = \mu S_t, \qquad b_t = \sigma S_t.
$$

These are the drift and diffusion coefficients of the stock price.

### Substituting into the general lemma

The general Itô's lemma for a function $f(t, X)$, where $f$ is the
price of a financial derivative whose value depends on the stock price
$S_t$ and time $t$, is

$$
df
=
\left(
\frac{\partial f}{\partial t}
+
a_t \frac{\partial f}{\partial x}
+
\frac{1}{2} b_t^2 \frac{\partial^2 f}{\partial x^2}
\right) dt
+
b_t \frac{\partial f}{\partial x} \, dB_t.
$$

Now set $X = S$, $a_t = \mu S$, and $b_t = \sigma S$. We get

$$
df
=
\left(
\frac{\partial f}{\partial t}
+
\mu S \frac{\partial f}{\partial S}
+
\frac{1}{2} \sigma^2 S^2 \frac{\partial^2 f}{\partial S^2}
\right) dt
+
\sigma S \frac{\partial f}{\partial S} \, dB_t.
$$

This is the **Itô's lemma for a function of a stock price**.

### Writing it compactly

Using subscripts for partial derivatives

$$
f_t = \frac{\partial f}{\partial t},
\qquad
f_S = \frac{\partial f}{\partial S},
\qquad
f_{SS} = \frac{\partial^2 f}{\partial S^2},
$$

the lemma becomes

$$
{
df
=
\left(
f_t
+
\mu S f_S
+
\frac{1}{2} \sigma^2 S^2 f_{SS}
\right) dt
+
\sigma S f_S \, dB_t.
}
$$

This is the equation from which the Black–Scholes PDE will be derived.

### Reading the equation

The change in $f$ has two parts:

- **A deterministic part** (the $dt$ coefficient):
  $$
  f_t + \mu S f_S + \frac{1}{2} \sigma^2 S^2 f_{SS}.
  $$
  This is the drift of $f$. It says: if the randomness were switched
  off, how would $f$ evolve over time?

- **A random part** (the $dB_t$ coefficient):
  $$
  \sigma S f_S.
  $$
  This is the diffusion of $f$. It says: how much does $f$ move when
  the Brownian shock $dB_t$ hits?
